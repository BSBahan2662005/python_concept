# File Operation-Read And Write Files

file handeling is a crucial part of any programming language.Python provides built-in functions and methods to read from and write to file,both text and binay.This lesson will cover the basic of file handeling,including reading and writting text files and binary files.

```python
# Read a whole file
# make sure that the file and the code file in same folder or directories

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
# read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line)
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) # strip() remove the newline character
```

```python
# writting a file (overwritting)

with open('example.txt','w') as file:
    file.write('Hello World!\n')
    file.write('this is a new line.')
```

```python
# write a file with out overwriting

with open('example.txt','a') as file:
    file.write("Append operation take place!")
```

```python
# writing list of line to a file
lines=['First line\n','Second line\n','Third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
# Binary file

# writing to a binary file
data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
# reading a binary file
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

```python
# read the content from a source file and write to a destinatio text file
with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
 
     
```

```python
# read the text file and count the number of lines , words, and characters.
# counting lines words and characters in a text file
def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines=file.readlines()
        line_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return line_count,word_count,char_count

file_path='example.txt'
lines,words , char=count_text_file(file_path)
print(f"Lines:{lines}, words:{words},characters:{char}")
```

the w+ mode in python is used to open a file for both reading and writing .if the file doesnot exist it will be created.if the file exists, it's content is truncate(i.e.,the file is overwritten)

```python
# writing and then reading a file
with open('example.txt','w+') as file:
    file.write("Hello World\n")
    file.write("This is a new line \n")

    # move the file cursor to the begining
    file.seek(0)

    # read the content of the file
    content=file.read()
    print(content)
```
