# 1️⃣ What is a Tuple?

A tuple in Python is a built-in data structure used to store multiple items in a single variable.

Key points:

Tuple is ordered

Tuple is immutable (cannot be changed after creation)

Allows duplicate values

Can store different data types

```python
t = (10, 20, 30, 40)

print(t)
print(type(t))
```

# Creating Tuples

Normal Tuple

```python
t = (1, 2, 3, 4)
print(t)
```

Tuple without parentheses

Python automatically treats comma-separated values as tuple.

```python
t = 1, 2, 3, 4
print(t)
```

# Single Element Tuple

Very important for beginners ⚠️

```python
t = (5,)     # correct
print(type(t))
```

```python
t = (5)      # this is integer
print(type(t))
```

# Tuple with Different Data Types

```python
t = (10, "python", 5.5, True)

print(t)
```

# Nested Tuple

Tuple inside another tuple.

```python
t = (1, 2, (10, 20, 30), 4)

print(t)
print(t[2])
```

# Acess tuple elements(index,slicing)

```python
# Tuple for demonstration
t = (10, 20, 30, 40, 50, 60, 70, 80)

print("Original Tuple:", t)

print("\n----- INDEXING -----")

# Positive indexing
print("t[0] =", t[0])      # first element
print("t[3] =", t[3])      # fourth element
print("t[7] =", t[7])      # last element using positive index

# Negative indexing
print("t[-1] =", t[-1])    # last element
print("t[-2] =", t[-2])    # second last element
print("t[-5] =", t[-5])    # 5th from end


print("\n----- BASIC SLICING -----")

# start : end
print("t[1:5] =", t[1:5])      # elements from index 1 to 4
print("t[2:6] =", t[2:6])      # elements from index 2 to 5

# start : end : step
print("t[1:7:2] =", t[1:7:2])  # skip every 2 elements


print("\n----- DEFAULT SLICING -----")

print("t[:] =", t[:])          # copy whole tuple
print("t[:4] =", t[:4])        # from start to index 3
print("t[3:] =", t[3:])        # from index 3 to end


print("\n----- STEP SLICING -----")

print("t[::2] =", t[::2])      # every second element
print("t[::3] =", t[::3])      # every third element


print("\n----- REVERSE SLICING -----")

print("t[::-1] =", t[::-1])    # reverse tuple
print("t[::-2] =", t[::-2])    # reverse with step 2


print("\n----- NEGATIVE SLICING -----")

print("t[-5:-1] =", t[-5:-1])  # slicing using negative indexes
print("t[-7:-2] =", t[-7:-2])
```

# Tuple is Immutable

Tuple cannot be modified.

```python
t = (10,20,30)

t[1] = 50
```

# Tuple Packing

Putting multiple values into a tuple.

```python
t = 10, 20, 30

print(t)
```

# Tuple Unpacking

Extract values from tuple into variables.

```python
t = (10,20,30)

a,b,c = t

print(a)
print(b)
print(c)
```

# Loop Through Tuple

```python
t = (10,20,30,40)

for i in t:
    print(i)
```

```python
t = (10,20,30,40)

for i in range(len(t)):
    print(t[i])
```

# Tuple Methods

Tuple has only 2 built-in methods.

count()

Counts how many times a value appears.

```python
t = (1,2,3,2,4,2,5)

print(t.count(2))
```

index()

Returns the first occurrence index.

```python
t = (10,20,30,40)

print(t.index(30))
```

# Built-in Functions Used with Tuple

```python
t = (10,20,30,40)

print(len(t))
print(max(t))
print(min(t))
print(sum(t))
```

# Converting List ↔ Tuple

List → Tuple

```python
l = [1,2,3,4]

t = tuple(l)

print(t)
```

Tuple → List

```python
t = (1,2,3,4)

l = list(t)

print(l)
```

Tuple Concatenation

```python
t1 = (1,2,3)
t2 = (4,5,6)

t3 = t1 + t2

print(t3)
```
