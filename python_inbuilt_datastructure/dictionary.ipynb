# A dictionary in Python is a built-in data structure used to store data in key–value pairs.

key : value

# Characteristics of Dictionary

Stores data in key–value pairs

Ordered (Python 3.7+)

Mutable (can change values)

Keys must be unique

Values can be duplicated

Keys must be immutable types (string, number, tuple)

```python
student = {
    "name": "Rakesh",
    "age": 20,
    "course": "Python"
}
```

```python
d = {"a":1, "b":2, "c":3}
print(d)
```

# Creating Dictionary
Method 1: Using {}

```python
d = {
    "name":"Rakesh",
    "age":21,
    "city":"Delhi"
}

print(d)
```

Method 2: Using dict()

```python
d = dict(name="Rakesh", age=21)

print(d)
```

# Accessing Dictionary Elements
Using Key

```python
student = {"name":"Rakesh","age":21}

print(student["name"])
```

Using get()

```python
student = {"name":"Rakesh","age":21}

print(student.get("age"))
```

dict["key"] → error if key missing

dict.get() → returns None

# Adding or Updating Elements

```python
d = {"a":1,"b":2}

# add new key
d["c"] = 3

# update value
d["a"] = 10

print(d)
```

# Looping Through Dictionary
Loop keys

```python
d = {"a":1,"b":2,"c":3}

for key in d:
    print(key)
```

Loop values

```python
for value in d.values():
    print(value)
```

Loop key and value

```python
for k,v in d.items():
    print(k,v)
```

# Dictionary Methods

keys()

Returns all keys.

```python
d = {"a":1,"b":2,"c":3}

print(d.keys())
```

values()

Returns all values.

```python
print(d.values())
```

items()

Returns key-value pairs.

```python
print(d.items())
```

get()

Gets value of key.

```python
print(d.get("a"))
```

update()

Update dictionary.

```python
d = {"a":1,"b":2}

d.update({"b":20,"c":30})

print(d)
```

pop()

Removes specific key.

```python
d = {"a":1,"b":2,"c":3}

d.pop("b")

print(d)
```

popitem()

Removes last inserted element.

```python
d = {"a":1,"b":2,"c":3}

d.popitem()

print(d)
```

clear()

Removes all elements.

```python
d = {"a":1,"b":2}

d.clear()

print(d)
```

copy()

Creates copy of dictionary.

```python
d1 = {"a":1,"b":2}

d2 = d1.copy()

print(d2)
```

setdefault()

Returns value if key exists otherwise creates key.

```python
d = {"a":1,"b":2}

d.setdefault("c",100)

print(d)
```

fromkeys()

Creates dictionary from keys.

```python
keys = ["a","b","c"]

d = dict.fromkeys(keys,0)

print(d)
```

# Built-in Functions with Dictionary

```python
d = {"a":10,"b":20,"c":30}

print(len(d))
print(max(d))
print(min(d))
print(sorted(d))
```

Membership Operators

```python
d = {"a":1,"b":2}

print("a" in d)
print("z" not in d)
```

# Nested Dictionary

Dictionary inside dictionary.

```python
students = {
    101: {"name":"Ravi","marks":80},
    102: {"name":"Aman","marks":90}
}

print(students[101]["name"])
```

# Dictionary Comprehension

```python
squares = {x: x*x for x in range(1,6)}

print(squares)
```
