# List Comprehension (Important)

What is List Comprehension?

List comprehension is a compact way to create a new list by processing elements of another iterable (list, range, tuple, etc.).

new_list = [expression for item in iterable if condition]

Manual Way (Traditional Loop)

```python
numbers = []

for i in range(1,6):
    numbers.append(i*i)

print(numbers)
```

List Comprehension

```python
numbers = [i*i for i in range(1,6)]

print(numbers)
```

```python
# manual way
even_numbers = []

for i in range(1,11):
    if i % 2 == 0:
        even_numbers.append(i)

print(even_numbers)
```

```python
# list comprehension
even_numbers = [i for i in range(1,11) if i%2==0]

print(even_numbers)
```

```python
# manual way modifing list elements
words = ["python","java","c++"]

upper_words = []

for w in words:
    upper_words.append(w.upper())

print(upper_words)
```

```python
# list comprehension
words = ["python","java","c++"]

upper_words = [w.upper() for w in words]

print(upper_words)
```

```python
# manual way only take square of the even number
numbers = [1,2,3,4,5,6]

result = []

for i in numbers:
    if i%2 == 0:
        result.append(i*i)

print(result)
```

```python
# list comprehension
numbers = [1,2,3,4,5,6]

result = [i*i for i in numbers if i%2==0]

print(result)
```

```python
# manual way print the list element which is even or which is odd
numbers = [1,2,3,4,5]

result = []

for i in numbers:
    if i%2 == 0:
        result.append("Even")
    else:
        result.append("Odd")

print(result)
```

```python
#  list comprehension
numbers = [1,2,3,4,5]

result = ["Even" if i%2==0 else "Odd" for i in numbers]

print(result)
```

# Nested Loop Example (Advanced)
Goal

Create pairs from two lists.

```python
list1 = [1,2,3]
list2 = ['A','B']

pairs = []

for i in list1:
    for j in list2:
        pairs.append((i,j))

print(pairs)
```

```python
list1 = [1,2,3]
list2 = ['A','B']

pairs = [(i,j) for i in list1 for j in list2]

print(pairs)
```

```python
# manual way list1 = [1,2,3] Filter students who passed.
marks = [35,80,55,20,90]

passed = []

for m in marks:
    if m >= 40:
        passed.append(m)

print(passed)
```

```python
# list comprehension
marks = [35,80,55,20,90]

passed = [m for m in marks if m >= 40]

print(passed)
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]

flat = []

for row in matrix:
    for num in row:
        flat.append(num)

print(flat)
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]

flat = [num for row in matrix for num in row]

print(flat)
```

```python
# remove negetive number
numbers = [10,-5,20,-3,30]

positive = []

for n in numbers:
    if n > 0:
        positive.append(n)

print(positive)
```

```python
numbers = [10,-5,20,-3,30]

positive = [n for n in numbers if n>0]

print(positive)
```

```python
# Extract Specific Column (Data Science Trick)
students = [["Ravi",85],["Aman",90],["Priya",78]]

marks = []

for student in students:
    marks.append(student[1])

print(marks)
```

```python
students = [["Ravi",85],["Aman",90],["Priya",78]]

marks = [student[1] for student in students]

print(marks)
```
