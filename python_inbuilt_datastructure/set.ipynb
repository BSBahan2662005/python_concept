# A set in Python is a built-in data structure used to store multiple unique elements.

Important Properties

Unordered

Unindexed

No duplicate values

Mutable (can add/remove items)

Written using {}

```python
s = {10, 20, 30, 40}
print(s)
print(type(s))
```

# Creating Sets

```python
# Normal set
s1 = {1,2,3,4}

# Set with duplicates (duplicates removed)
s2 = {1,2,2,3,3,4}

print(s1)
print(s2)
```

# Creating Empty Set

```python
s = {}
print(type(s))   # dictionary
# correct way to create a empty set
s = set()
print(type(s))
```

# Set Cannot Use Indexing

```python
s = {10,20,30}

print(s[0])
```

# Looping Through a Set

```python
s = {10,20,30,40}

for i in s:
    print(i)
```

# Adding Elements
add()

```python
s = {1,2,3}

s.add(4)

print(s)
```

update()

Adds multiple elements

```python
s = {1,2,3}

s.update([4,5,6])

print(s)
```

# Removing Elements
remove()

Removes element but gives error if not found

```python
s = {10,20,30}

s.remove(20)

print(s)
```

discard()

Removes element but no error if not found

```python
s = {10,20,30}

s.discard(40)

print(s)
```

pop()

Removes random element

bydefult the last elements of the set

```python
s = {10,20,30,40}

s.pop()

print(s)
```

clear()

Removes all elements

```python
s = {1,2,3}

s.clear()

print(s)
```

# Copy Set

```python
s1 = {1,2,3}

s2 = s1.copy()

print(s2)
```

# Set Operations (Very Important)

Union

Combine two sets

```python
A = {1,2,3}
B = {3,4,5}

print(A.union(B))
```

```python
print(A | B)
```

Intersection

Common elements

```python
A = {1,2,3}
B = {2,3,4}

print(A.intersection(B))
```

```python
print(A & B)
```

Difference

Elements in A not in B

```python
A = {1,2,3}
B = {2,3,4}

print(A.difference(B))
```

```python
print(A - B)
```

Symmetric Difference

Elements not common in both

```python
A = {1,2,3}
B = {3,4,5}

print(A.symmetric_difference(B))
```

```python
print(A ^ B)
```

# Set Relation Methods
issubset()

```python
A = {1,2}
B = {1,2,3}

print(A.issubset(B))
```

issuperset()

```python
A = {1,2,3}
B = {1,2}

print(A.issuperset(B))
```

isdisjoint()

Check if sets have no common elements

```python
A = {1,2}
B = {3,4}

print(A.isdisjoint(B))
```

# Built-in Functions Used With Sets

```python
s = {10,20,30,40}

print(len(s))
print(max(s))
print(min(s))
print(sum(s))
print(sorted(s))
```

# Convert List → Set

```python
l = [1,2,2,3,3,4]

s = set(l)

print(s)
```

# Convert Set → List

```python
s = {1,2,3}

l = list(s)

print(l)
```

# Membership Operators

```python
s = {10,20,30}

print(20 in s)
print(50 not in s)
```

# Frozen Set (Advanced Concept)

A frozenset is an immutable set.

You cannot add or remove elements.

```python
fs = frozenset([1,2,3,4])

print(fs)
```
