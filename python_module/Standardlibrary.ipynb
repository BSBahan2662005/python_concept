# 1️⃣ Create a File in Current Working Directory

```python
import os

file_name = "myfile.txt"

# Create file
with open(os.path.join(os.getcwd(), file_name), "w") as f:
    f.write("Hello, this file was created using os module!")

print("File created in:", os.getcwd())
```

os.getcwd() → gets the current working directory

os.path.join() → safely joins the path and filename

"w" → creates the file if it does not exist

# 3️⃣ Create Multiple Files Automatically

```python
import os

for i in range(5):
    file_path = os.path.join(os.getcwd(), f"file_{i}.txt")
    
    with open(file_path, "w") as f:
        f.write(f"This is file {i}")

print("5 files created successfully")
```

# 4️⃣ Create File Only If It Doesn't Exist

```python
import os

file_path = os.path.join(os.getcwd(), "data.txt")

if not os.path.exists(file_path):
    open(file_path, "w").close()
    print("File created")
else:
    print("File already exists")
```

```python
# Data Serialization
import json
data={'name':'Krish','age':25}

# converting dictionary in to json string
json_str=json.dumps(data)
print(json_str)
print(type(json_str))

# convertint json string into dictionary
past_data=json.loads(json_str)
print(past_data)
print(type(past_data))
```

```python
# csv
import csv

with open('example.csv',mode='w',newline='') as file:
   writer= csv.writer(file)
   writer.writerow(['name','age'])
   writer.writerow(['Bahan',20])

with open('example.csv',mode='r') as file:
   reader=csv.reader(file)
   for row in reader:
      print(row)
```
