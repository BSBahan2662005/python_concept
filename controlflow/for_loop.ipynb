```python
import json
import os

FILE_NAME = "bank_data.json"

# Load data from file
def load_data():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as f:
            return json.load(f)
    else:
        return {}

# Save data to file
def save_data(data):
    with open(FILE_NAME, "w") as f:
        json.dump(data, f, indent=4)


# Create new account
def create_account(data):
    username = input("Enter username: ")

    if username in data:
        print("User already exists!")
        return

    pin = input("Set 4-digit PIN: ")
    data[username] = {
        "pin": pin,
        "balance": 0,
        "transactions": []
    }

    save_data(data)
    print("Account created successfully!")


# Login system
def login(data):
    username = input("Enter username: ")
    pin = input("Enter PIN: ")

    if username in data and data[username]["pin"] == pin:
        print("Login successful!")
        user_menu(data, username)
    else:
        print("Invalid username or PIN")


# User menu after login
def user_menu(data, username):

    while True:

        print("\n1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Transaction History")
        print("5. Logout")

        choice = input("Enter choice: ")

        if choice == "1":
            print("Balance:", data[username]["balance"])

        elif choice == "2":
            amount = float(input("Enter amount to deposit: "))
            data[username]["balance"] += amount
            data[username]["transactions"].append(f"Deposited {amount}")
            save_data(data)
            print("Deposit successful")

        elif choice == "3":
            amount = float(input("Enter amount to withdraw: "))

            if amount <= data[username]["balance"]:
                data[username]["balance"] -= amount
                data[username]["transactions"].append(f"Withdrew {amount}")
                save_data(data)
                print("Withdrawal successful")
            else:
                print("Insufficient balance")

        elif choice == "4":
            print("\nTransaction History:")
            for t in data[username]["transactions"]:
                print("-", t)

        elif choice == "5":
            print("Logged out")
            break

        else:
            print("Invalid choice")


# Main program
def main():

    data = load_data()

    while True:

        print("\n===== BANK SYSTEM =====")
        print("1. Create Account")
        print("2. Login")
        print("3. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            create_account(data)

        elif choice == "2":
            login(data)

        elif choice == "3":
            print("Thank you for using the bank system")
            break

        else:
            print("Invalid option")


main()
```
