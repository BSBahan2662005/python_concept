# 1️⃣ What is a Lambda Function?

A lambda function is a small anonymous function.

Anonymous means it does not have a name.

It is used for short and simple operations.

# Normal Function

```python
def add(a, b):
    return a + b

print(add(5, 3))
```

# Lambda Version

```python
add = lambda a, b: a + b

print(add(5, 3))
```

# 2️⃣ Lambda Function Syntax

lambda arguments : expression

```python
square = lambda x: x * x
print(square(4))
```

# 3️⃣ Lambda with One Argument

```python
double = lambda x: x * 2

print(double(5))
```

# 4️⃣ Lambda with Multiple Arguments

```python
multiply = lambda a, b: a * b

print(multiply(4, 5))
```

# 5️⃣ Lambda Inside print()

```python
print((lambda x, y: x + y)(3, 7))
```

# 6️⃣ Lambda with map() (Very Common)

map() applies a function to every element in a list.

```python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x*x, numbers))

print(result)
```

# 7️⃣ Lambda with filter()

filter() selects elements based on a condition.

```python
numbers = [1,2,3,4,5,6,7,8]

even = list(filter(lambda x: x % 2 == 0, numbers))

print(even)
```

# 8️⃣ Lambda with sorted()

Used for custom sorting.

```python
students = [("Rakesh", 85), ("Amit", 90), ("Rahul", 78)]

sorted_students = sorted(students, key=lambda x: x[1])

print(sorted_students)
```

# 9️⃣ Lambda with Multiple Conditions

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(5))
print(check(10))
```

# 🔟 Lambda Inside Functions

```python
def operation(n):
    return lambda x: x * n

double = operation(2)

print(double(5))
```

# ⚠️ Limitations of Lambda

Lambda functions:

❌ Cannot have multiple statements

❌ Cannot use loops

❌ Cannot use complex logic
